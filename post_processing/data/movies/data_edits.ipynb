```python
users_fold1 = pd.read_csv("ml-1m_fold1_MultiVAE_test_recommendations.csv", names = ["user", "item", "score"])
compat = pd.read_csv("compat_1m_tail.csv", names = ["user", "item", "compatibility"])
```

```python
compat_tail = compat[compat["item"] == "women_writer_director"]
```

```python
missing = set(users_fold1["user"]) - set(compat_tail["user"])
print(len(missing), "users in fold1 but not in tail")
missing
```

```python
compat = pd.concat([compat, pd.DataFrame({"user": list(missing), "item": "women_writer_director", "compatibility": 0.0})], ignore_index=True)
```
